## Part D — Missing values

The real HDI file has **0 missing values**, so there is
nothing to fill in it. To learn the missing-value tools,
we make a labelled PRACTICE COPY with a few NaNs put in
by hand. The original `df` is never touched.



```python
print(df.isnull().sum().sum())
df_demo = df.copy()
df_demo.loc[0, 'HDI'] = np.nan
df_demo.loc[1, 'LifeExp'] = np.nan
df_demo.loc[2, 'GNI'] = np.nan
print(df_demo.isnull().sum())
```

```python
print(df_demo.dropna().shape)
```

```python
print(df_demo.dropna(subset=['HDI']).shape)
```

```python
df_demo.loc[5]=np.nan
print(df_demo)
```

```python
df_demo[5]=np.nan
print(df_demo.dropna(how='all').shape)
print(df_demo.shape)
```

```python
print(df_demo['HDI'].fillna(0.5).head(5).tolist())
```

```python
print(df_demo['HDI'].fillna(df_demo['HDI'].mean()).head(3).tolist())
```

```python
print(df_demo['HDI'].ffill().head(5).tolist())
```

```python
print(df_demo['HDI'].bfill().head(4).tolist())
```
